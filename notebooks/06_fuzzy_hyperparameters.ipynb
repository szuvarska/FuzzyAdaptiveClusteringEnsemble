```python
import itertools
DATASETS = ["wine", "glass", "ecoli"]
ALPHA_1_GRID = [0.4, 0.5, 0.6, 0.7, 0.8]
M_RANGE_GRID = [(1.2, 1.8), (1.5, 2.0), (1.5, 2.5)]

records = []
for a1, m_range in itertools.product(ALPHA_1_GRID, M_RANGE_GRID):
    tag = f"grid_a1_{a1}_m_{m_range[0]}_{m_range[1]}"
    cfg = ExperimentConfig(
        name=f"fuzzy_grid/{tag}",
        datasets=DATASETS,
        settings=["fixed", "variable"],
        methods=["Fuzzy-ACE"],
        n_runs=10,
        sampling_ratio=1.0,
        alpha_1=a1,
        fuzzy_ace_m_range=m_range,
        n_jobs=-1,
    )
    raw = run_experiment(cfg, resume=True, verbose=False)
    ok = raw[raw["status"] == "ok"]
    for (setting, ds), g in ok.groupby(["setting", "dataset"]):
        records.append({
            "alpha_1": a1,
            "m_range": str(m_range),
            "setting": setting,
            "dataset": ds,
            "ari_mean": g["ari"].mean(),
            "ari_std": g["ari"].std(),
            "nmi_mean": g["nmi"].mean(),
        })

grid_df = pd.DataFrame(records)
# Rank combinations by average ARI across the 3 datasets
summary = (
    grid_df.groupby(["setting", "alpha_1", "m_range"])
    .agg(avg_ari=("ari_mean", "mean"), avg_std=("ari_std", "mean"), avg_nmi=("nmi_mean", "mean"))
    .reset_index()
    .sort_values(["setting", "avg_ari"], ascending=[True, False])
)
display(summary)
```

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

PROJECT_ROOT = os.path.abspath(os.path.join(os.getcwd(), ".."))
GRID_DIR = os.path.join(PROJECT_ROOT, "results", "fuzzy_grid")

M_RANGES = [(1.5, 2.5), (1.5, 2.0), (1.2, 1.8), (1.05, 1.5)]
ALPHA_1 = 0.8

rows = []
for m_lo, m_hi in M_RANGES:
    tag = f"grid_a1_{ALPHA_1}_m_{m_lo}_{m_hi}"
    csv_path = os.path.join(GRID_DIR, tag, "raw_results.csv")
    df = pd.read_csv(csv_path)
    ok = df[df["status"] == "ok"]

    per_ds = (
        ok.groupby(["setting", "dataset"])
        .agg(
            ari_mean=("ari", "mean"), ari_std=("ari", "std"),
            nmi_mean=("nmi", "mean"), nmi_std=("nmi", "std"),
        )
        .reset_index()
    )
    avg = (
        per_ds.groupby("setting")
        .agg(
            avg_ari=("ari_mean", "mean"), avg_ari_std=("ari_std", "mean"),
            avg_nmi=("nmi_mean", "mean"), avg_nmi_std=("nmi_std", "mean"),
        )
        .reset_index()
    )
    avg["m_label"] = f"({m_lo}, {m_hi})"
    rows.append(avg)

plot_df = pd.concat(rows, ignore_index=True)
labels = [f"({lo}, {hi})" for lo, hi in M_RANGES]
x = np.arange(len(labels))

fig, axes = plt.subplots(1, 2, figsize=(10, 4), sharey=True)
for ax, (setting, title) in zip(axes, [("fixed", "Fixed $k$"), ("variable", "Variable $k$")]):
    sub = plot_df[plot_df["setting"] == setting].set_index("m_label").loc[labels]
    ax.errorbar(x - 0.05, sub["avg_ari"], yerr=sub["avg_ari_std"], marker="o", capsize=4, label="Avg ARI (± std)", linewidth=2)
    ax.errorbar(x + 0.05, sub["avg_nmi"], yerr=sub["avg_nmi_std"], marker="s", linestyle="--", capsize=4, label="Avg NMI (± std)", linewidth=2)
    ax.set_title(title)
    ax.set_xticks(x)
    ax.set_xticklabels(labels)
    ax.set_xlabel("Fuzzifier $m$ range")
    ax.grid(alpha=0.3)
    ax.legend()

axes[0].set_ylabel("Score (avg over Wine, Glass, Ecoli)")
fig.suptitle(f"Fuzzy-ACE Sensitivity to Fuzzifier Range ($\\alpha_1 = {ALPHA_1}$)")
fig.tight_layout()
fig.savefig(os.path.join(GRID_DIR, "m_sensitivity.png"), dpi=300, bbox_inches="tight")
plt.show()
```
