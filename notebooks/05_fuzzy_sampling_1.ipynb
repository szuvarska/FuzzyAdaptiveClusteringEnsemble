```python
import os
import sys
import pandas as pd

PROJECT_ROOT = os.path.abspath(os.path.join(os.getcwd(), ".."))
if PROJECT_ROOT not in sys.path:
    sys.path.insert(0, PROJECT_ROOT)

from src.experiments import ExperimentConfig, run_experiment
```

```python
DATASETS = ["iris", "wine", "breast_cancer"]

cfg = ExperimentConfig(
    name="fuzzy_sampling_1_0",
    datasets=DATASETS,
    settings=["fixed", "variable"],
    methods=["FCM", "Fuzzy-ACE"],
    n_runs=10,
    sampling_ratio=1.0,
    n_jobs=-1,
)

raw = run_experiment(cfg, resume=True, verbose=True)
```

```python
s_10 = pd.read_csv(os.path.join(PROJECT_ROOT, "results", "fuzzy_sampling_1_0", "summary.csv"))
s_07 = pd.read_csv(os.path.join(PROJECT_ROOT, "results", "notebook_test", "summary.csv"))

s_10 = s_10[s_10["method"].isin(["FCM", "Fuzzy-ACE"])].assign(sampling="1.0")
s_07 = s_07[
    (s_07["dataset"].isin(DATASETS)) & (s_07["method"] == "Fuzzy-ACE")
].assign(sampling="0.7")

comp = pd.concat([s_07, s_10], ignore_index=True)
comp["ARI"] = comp["ari_mean"].map("{:.3f}".format) + " ± " + comp["ari_std"].map("{:.3f}".format)
comp["NMI"] = comp["nmi_mean"].map("{:.3f}".format) + " ± " + comp["nmi_std"].map("{:.3f}".format)

display(
    comp[["setting", "dataset", "method", "sampling", "ARI", "NMI"]]
    .sort_values(["setting", "dataset", "method", "sampling"])
    .reset_index(drop=True)
)
```
